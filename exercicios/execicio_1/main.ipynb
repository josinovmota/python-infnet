```python
import csv
```

# Questão 1 - Carregar e exibir o conteúdo do CSV

```python
f = open("imdb_data.csv", "r")

parsed_csv = csv.reader(f, delimiter=";")
parsed_csv = list(parsed_csv)

for index in range(1, 6):
    print(parsed_csv[index])
```

# Questão 2 - Filmes e Séries com Maior Avaliação

```python
def best_movie_by_rating(csv):
    best_movie = csv[1]

    for row in csv[1:]:
        best_movie = row.copy() if float(row[5]) > float(best_movie[5]) else best_movie

    print(f"O melhor {best_movie[3]} é: {best_movie[2]}")

best_movie_by_rating(parsed_csv)
```

# Questão 3 - Contar Filmes, Séries e Mini Séries

```python
movie_count = 0
tvseries_count = 0
tvminiseries_count = 0

for row in parsed_csv[1:]:
    match row[3]:
        case 'movie':
            movie_count += 1
        case 'tvSeries':
            tvseries_count += 1
        case 'tvMiniSeries':
            tvminiseries_count += 1

print(f"A quantidade de Filmes é: {movie_count}\nA quantidade de Séries é: {tvseries_count}\nE a quantidade de Mini Séries é: {tvminiseries_count}")
```

# Questão 4 - Filmes/Séries Lançados antes de 2000

```python
movies_before_2000 = []

for row in parsed_csv[1:]:
    if int(row[-1]) < 2000:
        movies_before_2000.append(row)

list_best_movies_before_2000 = []

for index in range(10):
    list_best_movies_before_2000.append(movies_before_2000[index][2])

string_best_movies_before_2000 = ", ".join(item for item in list_best_movies_before_2000[:-1]) + f" e {list_best_movies_before_2000[-1]}"

print(f"Os melhores filmes lançados antes dos anos 2000 são: {string_best_movies_before_2000}")
```

# Questão 5 - Média de Avaliação por Tipo

```python
# Questão 3 - Contar Filmes, Séries e Mini Séries
movie_count = 0
tvseries_count = 0
tvminiseries_count = 0

movie_sum = 0
tvseries_sum = 0
tvminiseries_sum = 0

for row in parsed_csv[1:]:
    match row[3]:
        case 'movie':
            movie_count += 1
            movie_sum += float(row[5])
        case 'tvSeries':
            tvseries_count += 1
            tvseries_sum += float(row[5])
        case 'tvMiniSeries':
            tvminiseries_count += 1
            tvminiseries_sum += float(row[5])

movie_avg = movie_sum / movie_count
tvseries_avg = tvseries_sum / tvseries_count
tvminiseries_avg = tvminiseries_sum / tvminiseries_count

print(f"A média dos Ratings dos Filmes é: {str(movie_avg)[:3]}")
print(f"A média dos Ratings das Séries é: {str(tvseries_avg)[:3]}")
print(f"A média dos Ratings das Mini Séries é: {str(tvminiseries_avg)[:3]}")
```

# Questão 6 - Top 3 Filmes ou Séries Mais Votados

```python
parsed_csv_sorted_by_votes = []
parsed_csv_copy = parsed_csv.copy()[1:]

while parsed_csv_copy:
    minimum_vote = 10_000_000_000
    minimum_vote_row = None
    for row in parsed_csv_copy:
        if int(row[-2]) < minimum_vote:
            minimum_vote = int(row[-2])
            minimum_vote_row = row

    parsed_csv_sorted_by_votes.append(minimum_vote_row)
    parsed_csv_copy.remove(minimum_vote_row)

list_most_voted_films = []

for row in parsed_csv_sorted_by_votes[-1:-4:-1]:
    list_most_voted_films.append(row[2])

string_most_voted_films = ", ".join(str(item) for item in list_most_voted_films[:-1]) + f" e {list_most_voted_films[-1]}"

print(f"Os 3 filmes mais votados são: {string_most_voted_films}")
```
